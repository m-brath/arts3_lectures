The `calculate_absxsec` function calculates absorption cross sections and uses several keyword arguments as inputs. See the function definition in `lineshape_pyarts3.py` or the contextual help (`calculate_absxsec?`).

### 1) Absorption cross section and coefficient
Choose the $183\,\text{GHz}$-line of water vapor and perform calculations over a restricted frequency range for a number of different pressures between $10^5\,\text{Pa}$ and $10^1\,\text{Pa}$. Use the keyword arguments *fmin* and *fmax* to adjust the frequency range. Keep the temperature constant.

* How does the shape of the spectral lines change?

By now we investigated absorption in terms of the absorption cross-section
$\sigma$. Another widely used unit is the absorption coefficient
$\alpha$. It takes the number concentration $n$ of the absorber
into account:
$$ \alpha = n\cdot\sigma $$

* Use the ideal gas law to calculate the number concentration.
* How does the absorption coefficient in the line centre change, if pressure is changed?

```python
# Define parameters
species = "H2O"
temperature = 300
pressures = [101325, 10132, 1013]  # [Pa]

# Call ARTS to calculate absorption cross sections at each pressure
for pressure in pressures:
    freq, abs_xsec = calculate_absxsec(species, pressure, temperature,
                                       fmin=181e9, fmax=186e9, fnum=1000)

    fig, ax = plt.subplots()
    ax.plot(freq / 1e9, abs_xsec)
    ax.set_xlim(freq.min() / 1e9, freq.max() / 1e9)
    ax.set_ylim(bottom=0)
    ax.set_xlabel("Frequency [GHz]")
    ax.set_ylabel(r"Abs. cross section [$\sf m^2$]")
    ax.set_title(f"{tag2tex(species)}  p: {pressure/100:.0f} hPa  T: {temperature:.0f} K")
    plt.show()
```

### 2) Linewidth
The full-width at half maximum (FWHM) is a measure of the line width. Use the function `linewidth()`, which is imported from the `lineshape_pyarts3` module, to calculate the FWHM for a given absorption spectrum.

* Make a plot of this as a function of altitude in pressure coordinates for a microwave line and an infrared absorption line.
* Hint: Use a low pressure when selecting a line.

```python
# Use the linewidth() function to calculate the FWHM for each pressure.
# calculate_absxsec() computes only the isolated line-by-line (Voigt)
# contribution (the self-/cross-continuum is deliberately excluded).
# Pressure broadening makes the 1 bar line about an order of magnitude
# wider than the 10 mbar one (-- about 0.06 GHz vs. about 6 GHz), so we
# use a frequency window (+/-4 GHz) wide enough to resolve the broadest
# line.  The FWHM should scale roughly with pressure.
print("Pressure [hPa]\tLinewidth [GHz]")
for pressure in [101325, 10132, 1013]:
    freq, abs_xsec = calculate_absxsec("H2O", pressure, 300,
                                       fmin=178e9, fmax=188e9, fnum=4000)
    fwhm = linewidth(freq, abs_xsec)
    print(f"{pressure/100:.0f}\t\t{fwhm/1e9:.4f}")
```
